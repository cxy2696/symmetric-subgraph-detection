```python
# Real Test
folder_path = 'twitter'

print("demo.py imported successfully")
print("Working directory:", os.getcwd())
print("twitter/ folder files:", os.listdir(folder_path) if os.path.exists(folder_path) else "Don't find folder")
```

```python
# Load data from folder
file_num = 12831
test_file = f'{folder_path}/{file_num}.edges'

G = nx.read_edgelist(test_file, nodetype=int, create_using=nx.Graph())
print(f"Loaded graph: {G.number_of_nodes()} nodes, {G.number_of_edges()} edges")
print(f"Overall density: {nx.density(G):.4f}")
```

```python
# Run the function
result = detect_symmetric_subgraphs(G, mu=-1.0, density_threshold=0.8)
print("Detected symmetric subgraphs:", result)

# Ground-truth comparison
test_file_circle = f'{folder_path}/{file_num}.circles'

with open(test_file_circle, 'r') as f:
    circles = [line.strip().split('\t')[1:] for line in f if line.strip()]
print("Ground-truth circles:", circles)
```
